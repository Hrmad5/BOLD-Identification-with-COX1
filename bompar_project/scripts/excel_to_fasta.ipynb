# Converts sequences from an Excel file to a FASTA file.

### Parameters:
### excel_path str: path to the input excel file
### fasta_path str: path to the output fasta file
### id_col str: sequence's ID
### seq_col str: column with sequences
### sheet_name str or int: name of the sheet or its index, default is 0

```python
import pandas as pd
from Bio import SeqIO
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
import sys

def excel_to_fasta(excel_path, fasta_path, id_col, seq_col, sheet_name=0):

    try:
        df = pd.read_excel(excel_path, sheet_name=sheet_name)

        records = []

        for index, row in df.iterrows():
            seq_id = str(row[id_col])
            sequence = str(row[seq_col])

            if not seq_id or not sequence:
                print(f"Skipping row {index + 2} in {excel_path}: Missing ID or sequence.")
                continue
            record = SeqRecord(Seq(sequence), id=seq_id, description="")
            records.append(record)

        if records:
            SeqIO.write(records, fasta_path, "fasta")
            print(f"Successfully wrote {len(records)} sequences to {fasta_path}.")
        else:
            print("No valid sequences found to write.")
        
    except FileNotFoundError:
        print(f"Error: The file {excel_path} was not found.")
    except KeyError as e:
        print(f"Error: Column {e} not found in the Excel file. Please check your column names.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

if __name__ == "__main__":
    INPUT_EXCEL = "../data/sequence_data.xlsx"
    OUTPUT_FASTA = "../data/sequence_data.fasta"
    ID_COLUMN = "Sample_ID"
    SEQUENCE_COLUMN = "Sequence"
    SHEET = "Sheet1"

    excel_to_fasta(INPUT_EXCEL, OUTPUT_FASTA, ID_COLUMN, SEQUENCE_COLUMN, SHEET)
```
